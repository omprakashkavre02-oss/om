# 🏠 House Price Prediction using Linear Regression

This notebook explores the housing dataset, preprocesses features, trains a Linear Regression model, evaluates it, and demonstrates a prediction.

```python
import pandas as pd
import numpy as np
from sklearn.model_selection import train_test_split
from sklearn.compose import ColumnTransformer
from sklearn.pipeline import Pipeline
from sklearn.impute import SimpleImputer
from sklearn.preprocessing import OneHotEncoder, StandardScaler
from sklearn.linear_model import LinearRegression
from sklearn.metrics import mean_absolute_error, mean_squared_error, r2_score
```

```python
df = pd.read_csv('../data/housing.csv')
df.head()
```

```python
target='price'
X=df.drop(columns=[target])
y=df[target]
X_train,X_test,y_train,y_test=train_test_split(X,y,test_size=0.20,random_state=42)
num=X_train.select_dtypes(include=['int64','float64']).columns.tolist()
cat=X_train.select_dtypes(include=['object','category']).columns.tolist()
pre=ColumnTransformer([('num',Pipeline([('imp',SimpleImputer(strategy='median')),('scale',StandardScaler())]),num),('cat',Pipeline([('imp',SimpleImputer(strategy='most_frequent')),('onehot',OneHotEncoder(handle_unknown='ignore'))]),cat)])
```

```python
model=Pipeline([('preprocessor',pre),('regressor',LinearRegression())])
model.fit(X_train,y_train)
pred=model.predict(X_test)
mse=mean_squared_error(y_test,pred)
print('MAE:',mean_absolute_error(y_test,pred))
print('MSE:',mse)
print('RMSE:',np.sqrt(mse))
print('R2:',r2_score(y_test,pred))
```

```python
sample=pd.DataFrame([{'area_sqft':1500,'bedrooms':3,'bathrooms':2,'stories':2,'parking':2,'location':'Urban','age_years':5}])
print(f'Predicted price: ${model.predict(sample)[0]:,.2f}')
```
